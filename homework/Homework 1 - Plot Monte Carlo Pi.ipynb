MC function

```python
def MC_pi(N_vals):
    diff = []

    for N in N_vals:
        x = np.random.uniform(-1, 1 , N)
        y = np.random.uniform(-1, 1 , N)

        N_hits = np.sum(x*x + y*y < 1)

        pi_mc = 4 * N_hits / N       
        diff.append(abs(pi_mc - np.pi))

    return np.array(diff)
```
