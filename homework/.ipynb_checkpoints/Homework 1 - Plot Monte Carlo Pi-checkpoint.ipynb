Importing Libraries 

```python
import numpy as np
import matplotlib.pyplot as plt 
```

Setting Range

```python
errors = []
N_vals = np.arange(10, 1001)

```

MC function

```python
def MC_pi(N_vals):
    errors = []

    for N in N_vals:
        x = np.random.uniform(-1, 1)
        y = np.random.uniform(-1, 1)

        N_hits = np.sum(x*x + y*y < 1)

        pi_mc = 4 * N_hits / N       
        errors.append(abs(pi_mc - np.pi))

    return np.array(errors)
```

Calling MC function 

```python
errors = MC_pi(N_vals) 
```

Plot Routine 

```python

plt.figure(figsize=(16, 4))
plt.scatter(N_vals, errors)
plt.xlabel('Number of Samples')
plt.ylabel('Variation from Pi')
plt.title("Monte Carlo Pi Approximation ")
plt.show()
```
