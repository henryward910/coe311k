```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def Taylor_Sin(x, terms):
    s = 0.0
    for k in range (terms):
        s += ((-1)**k) * (x**(2 * k + 1)) / math.factorial(2 * k + 1)    
    return s
    
```

```python
x_vals = np.arange(0, 10.1 , 0.1)

Ts_sin_1 = np.array([Taylor_Sin(x, 1) for x in x_vals])
Ts_sin_2 = np.array([Taylor_Sin(x, 2) for x in x_vals])
Ts_sin_3 = np.array([Taylor_Sin(x, 3) for x in x_vals])
Ts_sin_4 = np.array([Taylor_Sin(x, 4) for x in x_vals])


sin_true = np.sin(x_vals)


```

Plot to show difference in approximate based off the number of terms.

```python
d = np.linspace(0, 10, 101)
plt.figure(figsize = (12, 4))
plt.plot(d, np.sin(d), label = " Sin(x)")
plt.plot(d, Ts_sin_1, label = " 1 Term")
plt.plot(d, Ts_sin_2, label = " 2 Term")
plt.plot(d, Ts_sin_3, label = " 3 Term")
plt.plot(d, Ts_sin_4, label = " 4 Term")
plt.ylim(-1, 2)
plt.xlabel("x")
plt.ylabel("y.")
plt.title("Taylor Series Sin(x) 1-4 Terms ")
plt.legend()
plt.show()
```

Error Calculator:

```python
AE_1 = np.abs(sin_true - Ts_sin_1)
AE_2 = np.abs(sin_true - Ts_sin_2)
AE_3 = np.abs(sin_true - Ts_sin_3)
AE_4 = np.abs(sin_true - Ts_sin_4)
```

```python
plt.figure(figsize = (12, 4))
plt.plot(x_vals, AE_1, label = "1-Term Error")
plt.plot(x_vals, AE_2, label = "2-Term Error")
plt.plot(x_vals, AE_3, label = "3-Term Error")
plt.plot(x_vals, AE_4, label = "4-Term Error")
plt.xlabel("x")
plt.ylabel("Absolute Error")
plt.title("Absolute Error vs. Taylor Series Sin(x)")
plt.legend()
plt.show()
```

As more terms are included into the Taylor Approx. The error decreases. But as the second plot demonstrates the larger the degree of polynomial the more exponential the difference becomes. This is especially true at larger x values. This is because Taylor polynomials provide the most accurate approximations near the expansion point.
