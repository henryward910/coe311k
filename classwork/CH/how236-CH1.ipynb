```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*np.pi, 100)
```

```python
plt.figure(figsize=(14, 12)) 
plt.subplot(2, 2, 1) 
plt.plot(x, np.sin(x)**2, 'r-x', label='Sin^2(x)')
plt.plot(x, abs(np.sin(x)), 'g-^', label='|Sin(x)|')
plt.legend() 
plt.xlabel('Rads')
plt.ylabel('Amplitude')
plt.title('Sin^2 and Cos Waves')
plt.subplot(2, 2, 2)
plt.plot(np.sin(x), np.cos(x), 'g', label='(sin(x),cos(x))')
plt.legend() 
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Circle')
plt.subplot(2, 2, 3)
r = x
plt.plot(r*np.cos(x), r*np.sin(x), label='F(x)')
plt.title("Spiral")
plt.axis('equal')
plt.legend() 
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Sin and Cos Waves')
plt.subplot(2, 2, 4)
k = np.random.rand(1000)
j = np.random.rand(1000)
size = np.random.rand(1000) * 50
color = np.random.rand(1000)
plt.scatter(k, j, size, color)
plt.colorbar()
plt.show()
```
