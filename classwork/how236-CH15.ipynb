```python
import numpy as np
import matplotlib.pyplot as plt

x_start, x_end = 0, 2
step_sizes = np.arange(0.01, 1.05, 0.05)

def f(x, y):
    return -2 * y

# Exact solution for reference
x_ref = np.linspace(x_start, x_end, 1000)
y_exact = np.exp(-2 * x_ref)

plt.figure(figsize=(12, 6))

for h in step_sizes:
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1

    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    # Exact solution at same x points
    y_exact_at_x = np.exp(-2 * x_values)

    # Error at each point vs exact
    errors = np.abs(y_values - y_exact_at_x)
    total_error = np.sum(errors)

    print(f"h = {h:.2f} | point-wise errors: {np.round(errors, 5)} | total accumulated error: {total_error:.5f}")

```
