```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Data points 

X = np.array([1, 2, 3, 4, 5])

Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)

A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
```

```python
plt.scatter(X, Y)


x_line = np.linspace(min(X), max(X), 100)
y_line = m * x_line + c

plt.plot(x_line, y_line, color='r',label=f'y = {m:.2f}x + {c:.2f}')
plt.legend()
plt.show()
plt.show()
```

```python
v = 10 * np.random.rand(5) / 2
u = 10 * np.random.rand(5) / 2
```

```python
A1 = np.vstack([v, np.ones(len(v))]).T
```

```python
m1, c1 = np.linalg.lstsq(A1, Y, rcond=None)[0]

print(f"Slope: {m1}, Intercept: {c1}")
```

```python
plt.scatter(u,v)



x1_line = np.linspace(min(X), max(X), 100)
y1_line = m1 * x1_line + c1

plt.plot(x1_line, y1_line, color='b', label=f'y = {m1:.2f}x + {c1:.2f}')
plt.legend()
plt.show()
plt.show()
```

You can see now that due to the spread of the random points the fit is a lot more loose and is a far worse determiner for evaluating the best fit

```python
m_true = 1.5
b_true = -1
x = np.arange(0.0, 1.01, 0.01)
```

```python
rng = np.random.default_rng(42)     
noise_amp = 0.5                    
noise = rng.uniform(-noise_amp, noise_amp, size=x.shape)

y = m_true * x + b_true + noise

A = np.vstack([x, np.ones_like(x)]).T
m_fit, b_fit = np.linalg.lstsq(A, y, rcond=None)[0]

y_true = m_true * x + b_true
y_fit  = m_fit * x + b_fit
```

```python
plt.scatter(x, y, s=15, label= "Noisy data")
plt.plot(x, y_true, linewidth=2, label=f"True: y = {m_true:.2f}x + {b_true:.2f}")
plt.plot(x, y_fit, linewidth=2, label=f"Fit:   y = {m_fit:.2f}x + {b_fit:.2f}")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()

print(f"True slope/intercept: {m_true:.5f}, {b_true:.5f}")
print(f"Fit  slope/intercept: {m_fit:.5f}, {b_fit:.5f}")
```
