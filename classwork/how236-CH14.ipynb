```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = np.sin          
f_true = np.cos

a, b = 0, 2 * np.pi  #
N = 100             

x = np.linspace(a, b, N)
h = (b - a) / (N - 1)


df_central = (f(x + h) - f(x - h)) / (2 * h)

df_forward = (f(x + h) - f(x)) / h

df_backward = (f(x) - f(x - h)) / h

df_true = f_true(x)

```

```python
"""
Asked Claude to help and plot the graphs

"""





fig, axes = plt.subplots(2, 2, figsize=(13, 9), sharey=True)
fig.suptitle("Numerical Differentiation of sin(x)  [h = {:.4f}]".format(h),
             fontsize=15, fontweight="bold", y=1.01)

plot_cfg = [
    (df_true,     "True Derivative  cos(x)",  "#2196F3", "solid"),
    (df_forward,  "Forward Difference",        "#FF5722", "dashed"),
    (df_backward, "Backward Difference",       "#4CAF50", "dashed"),
    (df_central,  "Central Difference",        "#9C27B0", "dashed"),
]

for ax, (data, label, color, style) in zip(axes.flat, plot_cfg):
    ax.plot(x, df_true, color="lightgrey", linewidth=2.5,
            label="True  cos(x)", zorder=1)
    ax.plot(x, data, color=color, linewidth=2,
            linestyle=style, label=label, zorder=2)

    ax.set_title(label, fontsize=12, fontweight="bold", color=color)
    ax.set_xlabel("x")
    ax.set_ylabel("dy/dx")
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)
    ax.set_xlim(a, b)
    ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
    ax.set_xticklabels(["0", "π/2", "π", "3π/2", "2π"])

plt.tight_layout()
plt.show()
```
