```python
import numpy as np
import time 
import matplotlib.pyplot as plt
```

```python
A = np.array([[10 , -1 , 2 , 0 , 0],
              [-1 , 11, -1 , 3 , 0],
              [2 , -1 , 10 , -1 , 0],
              [0 , 3 , -1 , 8 , -2],
              [0 , 0, 0 , -2 , 9]])

b = np.array([14, 30, 26 , 25, 37])

x0 = np.zeros_like(b)

tol = 1e-6

max_iterations = 100
```

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    start_time = time.perf_counter()
    n = len(b)
    x = x0.copy()
    print(f"Initial Guess: x = {x}")
    for k in range(max_iterations):
        x_new = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        
    # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            duration = time.perf_counter() - start_time
            return x_new, (k), duration
    
        x = x_new
       
        
        duration = time.perf_counter() - start_time
             
    raise RuntimeError(f"Gauss-Seidel method did not converge after {max_iterations} iterations.")
```

```python
try:
    solutions, iterations, seconds = gauss_seidel(A, b, x0, tol, max_iterations)
    print("Solution:", solutions)
    print("Iterations:", iterations)
    print("Seconds:", seconds)
except RuntimeError as e:
    print(e)
```

```python
def jacobi(A, b, x0, tol, max_iterations):
    start_time = time.perf_counter()
    n = len(b)
    x = x0.copy()
    print(f"Initial Guess: x = {x}")
    
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        
        
# Check for convergence
            if np.linalg.norm(x_new - x, ord=np.inf) < tol:
                duration = time.perf_counter() - start_time
                return x_new, (k+1), duration
        
        x = x_new


    raise RuntimeError(
    f"Jacobi method did not converge after {max_iterations} iterations.")
```

```python
try:
    solutions, iterations, seconds = jacobi(A, b, x0, tol, max_iterations)
    print("Solution:", solutions)
    print("Iterations:", iterations)
    print("Seconds:", seconds)
except RuntimeError as e:
    print(e)
```
