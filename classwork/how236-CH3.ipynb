```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python

def MC_pi(N):

    x = np.random.uniform(-1, 1 , N)
    y = np.random.uniform(-1, 1 , N)

    N_hits = np.sum(x*x + y*y < 1)

    pi_mc = 4 * N_hits / N       
   

    return pi_mc
```

```python
TE100 = []
RE100 = []

for i in range(100):
    pi_calc = MC_pi(1000)
    TE100.append(abs(np.pi - pi_calc))
    RE100.append(abs(np.pi - pi_calc) / np.pi)
```

```python
plt.plot (range(100) , TE100, label='Absolute Error')
plt.plot (range(100) , RE100, label='Relative Error')
plt.xlabel('Number of Trials')
plt.ylabel('Error Magnitude')
plt.legend()
plt.show()
```
