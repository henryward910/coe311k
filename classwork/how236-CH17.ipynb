```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
k = 1.0
x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0

h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]
```

```python



plt.figure(figsize=(12, 5))

for h in h_values:
    
    A = np.array([[1,    h  ],
                  [-k*h, 1  ]])

   
    eigenvalues = np.linalg.eigvals(A)
    max_magnitude = np.max(np.abs(eigenvalues))

    print(f"h = {h}")
    if max_magnitude <= 1:
        print("  Solution is Stable")
    else:
        print("  Solution may not be Entirely Stable ;)")
    print(f"  Eigenvalue magnitudes: {np.abs(eigenvalues)}\n")


    x_vals = np.arange(x_start, x_end + h, h)
    y_vals = np.zeros(len(x_vals))
    v_vals = np.zeros(len(x_vals))
    y_vals[0] = y_0
    v_vals[0] = v_0

    for i in range(1, len(x_vals)):
        y_n = y_vals[i - 1]
        v_n = v_vals[i - 1]
        y_vals[i] = y_n + h * v_n
        v_vals[i] = v_n + h * (-k * y_n)

    plt.plot(x_vals, y_vals, label=f'h={h}')

x_exact = np.linspace(x_start, x_end, 1000)
plt.plot(x_exact, np.cos(np.sqrt(k) * x_exact), 'k--', lw=2, label='Exact')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title(f"d²y/dx² = -ky, k={k}")
plt.ylim(-5, 5)
plt.legend()
plt.grid(True)
plt.show()
```
