```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

def exact(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

h_values = [0.1, 0.01, 0.001, 0.00001, 0.0000001]
t_end = 0.1          # all interesting behavior is here

plt.figure(figsize=(12, 5))

t_exact = np.linspace(0, t_end, 10000)
plt.plot(t_exact, exact(t_exact), 'k--', lw=2, label='Analytical')

for h in h_values:
    t_values = np.arange(0, t_end + h, h)
    y_values = np.zeros(len(t_values))

    for i in range(1, len(t_values)):
        t_n = t_values[i - 1]
        y_n = y_values[i - 1]
        y_values[i] = np.clip(y_n + h * f(t_n, y_n), -1e10, 1e10)

    plt.plot(t_values, y_values, label=f'h={h}', alpha=0.8)

plt.xlabel('t')
plt.ylabel('y(t)')
plt.title("Euler's Forward Method  |  dy/dt = -1000y + 3000 - 2000e^(-t)")
plt.ylim(-1, 5)
plt.grid(True)
plt.show()
```
